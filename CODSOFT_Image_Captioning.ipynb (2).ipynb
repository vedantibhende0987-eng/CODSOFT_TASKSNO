```python

from transformers import pipeline

# Load the pre-trained image captioning model
captioner = pipeline(
    "image-to-text",
    model="Salesforce/blip-image-captioning-base"
)

print("Image Captioning AI is ready!")
```

```python

from google.colab import files
from PIL import Image
from IPython.display import display

# Select a picture from your computer
uploaded = files.upload()

# Get the uploaded picture
image_path = next(iter(uploaded))

# Open and display the picture
image = Image.open(image_path).convert("RGB")
display(image)

print("Picture uploaded successfully!")

```

```python

# Generate a caption for the uploaded picture
result = captioner(image)

# Extract the generated sentence
caption = result[0]["generated_text"]

print("IMAGE CAPTION:")
print(caption)

```
